### Reshaping, Indexing, Element-wise Operations, and Matrix Multiplication

This laboratory activity explores fundamental operations in PyTorch tensors, including tensor creation, data type conversion, reshaping, indexing, element-wise operations, and matrix multiplication. NumPy is initially used to generate a reproducible array, which is then converted into a PyTorch tensor for further manipulation.

The activity also demonstrates the importance of tensor shape compatibility when performing matrix multiplication and highlights how setting a random seed makes computational results reproducible.

## 1. Standard Imports

```python
import numpy as np
import torch
```

## 2. `set_seed()` Function

Seeds both NumPy's and PyTorch's random number generators so that random operations later in the
notebook are reproducible. It takes an `int` and returns nothing (`None`).

The `set_seed()` function initializes the random number generators of both NumPy and PyTorch using the same seed value.

The parameter is annotated as `int`, so that the function expects an integer seed. The return type is annotated as `None` because the function does not produce a value; instead, its purpose is to establish a fixed state for subsequent random operations.

```python
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)
    return None
```

## 3. Create `arr`: 6 Random Integers in [0, 5), Seeded with 42

`np.random.randint(low, high, size)` draws integers from `low` (inclusive) to `high` (exclusive).


The seed is first set to 42 to ensure that the generated values are reproducible. The `np.random.randint()` function generates six random integers where 0 is inclusive and 5 is exclusive.

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6)

print("arr =", arr)
print("dtype =", arr.dtype)
```

The array has six elements and uses the `int32` data type. This array serves as the starting point for creating the PyTorch tensor used throughout the succeeding tasks.

## 4. Create Tensor `x` From `arr`

```python
x = torch.from_numpy(arr)
print(x)
print(x.dtype)
```

The `torch.from_numpy()` function converts the NumPy array into a PyTorch tensor.

The values remain unchanged during the conversion. Since the original NumPy array has an int32 data type, the resulting tensor also has the torch.int32 data type. At this stage, x is a one-dimensional tensor containing six elements.

## 5. Change `x`'s dtype From `int32` to `int64`

```python
x = x.to(torch.int64)

print(x)
print(x.dtype)
```

The `.to(torch.int64)` operation converts the tensor from `int32` to `int64`.

The numerical values remain the same because changing the data type does not alter the values themselves. Instead, it changes how those values are represented and stored. This conversion is relevant because PyTorch commonly uses `int64` for integer-based tensor operations, particularly when integer tensors are used as indices or labels.

## 6. Reshape `x` Into a 3x2 Tensor

```python
x = x.reshape(3, 2)
print(x)
print(x.shape)
```

The original tensor contains six elements, which can be rearranged into a matrix with 3 rows and 2 columns because:

$$ 3 \times 2 = 6 $$

The reshape(3, 2) operation changes the structure of the tensor without changing the order or values of its elements.

The resulting tensor is:

$$ x = \begin{bmatrix} 3 & 4\\ 2 & 4\\ 4 & 1 \end{bmatrix} $$

Its shape is therefore `torch.Size([3, 2]).`

This reshaping is particularly important for the succeeding matrix multiplication task because the dimensions of x determine the compatible dimensions required for tensor y.

## 7. Return the Right-Hand Column of `x`

```python
right_column = x[:, 1]
print(right_column)
```

The expression `x[:, 1]` uses PyTorch indexing to select the right-hand column of the tensor.

- `:` selects all rows.
- `1` selects the second column, since Python uses zero-based indexing.

Therefore, from

$$ \begin{bmatrix} 3 & 4\\ 2 & 4\\ 4 & 1 \end{bmatrix} $$

the selected values are:

$$ [4,4,1] $$

This operation demonstrates how tensor indexing can be used to extract specific portions of multidimensional data without modifying the original tensor.

## 8. Return a Tensor of Squared Values of `x`, Without Changing `x`

```python
x_squared = torch.square(x)

print("x squared:")
print(x_squared)

print("\nx (unchanged):")
print(x)
```

The `torch.square()` function performs an element-wise squaring operation. Each element of x is multiplied by itself:

$$ x^2 = \begin{bmatrix} 3^2 & 4^2\\ 2^2 & 4^2\\ 4^2 & 1^2 \end{bmatrix} = \begin{bmatrix} 9 & 16\\ 4 & 16\\ 16 & 1 \end{bmatrix} $$

The result is stored in a new tensor called `x_squared.`

Importantly, `x` itself remains unchanged. This demonstrates the distinction between creating a new tensor from an operation and directly modifying the original tensor.

## 9. Create Tensor `y`: Same Number of Elements as `x`, Shaped to Allow `matmul(x, y)`

```python
set_seed(42)
y = torch.randint(0, 5, (2, 3))

print(y)
print(y.shape)
```

Tensor `y` is generated directly using PyTorch's `torch.randint()` function. The values are restricted to the range 0 to 4, with 5 as the exclusive upper bound.

The tensor contains six elements, matching the number of elements in x:

$$ 2 \times 3 = 6 $$

The shape of `y` is deliberately chosen as `2 × 3` rather than another six-element arrangement because it must be compatible with matrix multiplication.

The dimensions of the two tensors are:

$$ x: (3 \times 2) $$ $$ y: (2 \times 3) $$

For matrix multiplication, the inner dimensions must be equal:

$$ (3 \times \mathbf{2})(\mathbf{2} \times 3) $$

Since both inner dimensions are 2, the multiplication is valid. The resulting matrix will have the outer dimensions:

$$ 3×3 $$

## 10. Matrix Product of `x` and `y`

```python
product = torch.matmul(x, y)

print(product)
print(product.shape)
```

The matrix product is calculated `using torch.matmul().` Given:

$$ x = \begin{bmatrix} 3 & 4\\ 2 & 4\\ 4 & 1 \end{bmatrix} $$

and

$$ y = \begin{bmatrix} 2 & 2 & 1\\ 4 & 1 & 0 \end{bmatrix} $$

the resulting matrix is:

$$ xy = \begin{bmatrix} 22 & 10 & 3\\ 20 & 8 & 2\\ 12 & 9 & 4 \end{bmatrix} $$

For example, the first element is obtained by multiplying the first row of `x` by the first column of `y`:

$$ (3)(2) + (4)(4) = 6 + 16 = 22 $$

The same row-by-column process is applied to every element of the resulting matrix. The resulting shape is 3 × 3, which follows the matrix multiplication rule:

$$ (3 \times 2)(2 \times 3) = 3 \times 3 $$

This demonstrates why selecting the correct shape for y was necessary for the matrix multiplication operation.
