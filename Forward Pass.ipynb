# **Laboratory Task 2 — Forward Pass and Error Computation**

**Instruction:** Perform a single forward pass and compute for the error.

**Network architecture:** 3 inputs &rarr; 2 hidden units (ReLU) &rarr; 1 output unit (ReLU)

| Given | Value |
|---|---|
| Input `x` | `[1, 0, 1]` |
| Target `y` | `[1]` |
| Activation `f` | `max(0, Z_n)` (ReLU) |

**Hidden unit weights** (3 inputs &rarr; 2 hidden units)

| | Hidden Unit 1 | Hidden Unit 2 |
|---|---|---|
| from x1 | w11 = 0.2 | w12 = -0.3 |
| from x2 | w13 = 0.4 | w14 = 0.1 |
| from x3 | w15 = -0.5 | w16 = 0.2 |

**Output unit weights:** w21 = -0.3, w22 = -0.2

**Biases:** θ1 = -0.4 (hidden 1), θ2 = 0.2 (hidden 2), θ3 = 0.1 (output)


## Step 0 — Set Up Inputs, Weights, and Biases

```python
import numpy as np

# Inputs
x = np.array([1, 0, 1], dtype=float)
y = np.array([1], dtype=float)

# Activation function: ReLU
def relu(z):
    return np.maximum(0, z)

# Hidden unit weights
W1 = np.array([
    [0.2, -0.3],   # w11, w12  (from x1)
    [0.4,  0.1],   # w13, w14  (from x2)
    [-0.5, 0.2]    # w15, w16  (from x3)
])
b1 = np.array([-0.4, 0.2])   # theta1, theta2

# Output unit weights 
W2 = np.array([
    [-0.3],   # w21
    [-0.2]    # w22
])
b2 = np.array([0.1])   # theta3

print("x  =", x)
print("y  =", y)
print("W1 =\n", W1)
print("b1 =", b1)
print("W2 =\n", W2)
print("b2 =", b2)
```

## Step 1 — Hidden Layer: Net Input and Activation

$$Z_{h1} = x_1 w_{11} + x_2 w_{13} + x_3 w_{15} + \theta_1 \qquad
Z_{h2} = x_1 w_{12} + x_2 w_{14} + x_3 w_{16} + \theta_2$$


```python
z1 = x.dot(W1) + b1
a1 = relu(z1)

print("z1 (hidden net input)   :", z1)
print("a1 (hidden activation)  :", a1)
```

## Step 2 — Output Layer: Net Input and Activation

$$Z_o = a_{h1} w_{21} + a_{h2} w_{22} + \theta_3 \qquad \hat{y} = \max(0, Z_o)$$


```python
z2 = a1.dot(W2) + b2
a2 = relu(z2)

print("z2 (output net input)      :", z2)
print("a2 / y_hat (predicted output):", a2)
```

## Step 3 — Error Computation

$$E = \frac{1}{2}(y-\hat{y})^2$$

```python
error = 0.5 * np.sum((y - a2) ** 2)
print("Error:", error)
```

## Analysis & Discussion

- The network's prediction (**ŷ = 0.08**) is far from the target (**y = 1**), producing a large squared "
error of **0.4232**. This is expected behavior for a network with randomly-initialized weights that has "
not yet been trained.
- **Hidden unit 1 is "dead"** for this particular input: its net input (-0.7) falls below zero, so ReLU "
clips it to 0. It provides no signal to the output layer on this forward pass.
- **Hidden unit 2 is active** (0.1) and is currently the only path carrying information from input to "
output (this means the output almost entirely depends on `w22` and `θ3` for this sample.)
- Because the error is large and one unit is inactive, a single training step (gradient descent) will "
mostly adjust the weights connected to the *active* hidden unit. This is explored quantitatively in "
**Laboratory Task 3**, where backpropagation is used to compute these updates.
